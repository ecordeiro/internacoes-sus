# Internações da ANS

Saúde suplementar, padrão TISS. Não é o SUS. Uma execução baixa **tudo o que a ANS já publicou**: os 27 estados, de 2015 até o último ano no FTP, nos três arquivos.

| Tipo | Pasta | O que é uma linha |
| --- | --- | --- |
| `cons` | `dados/ans/parquet/cons` | uma internação |
| `det` | `dados/ans/parquet/det` | um item dessa internação: procedimento, medicamento, material ou diária |
| `rem` | `dados/ans/parquet/rem` | o modelo de remuneração; a ANS passou a publicar em 2023 |

A chave entre os três é `ID_EVENTO_ATENCAO_SAUDE`. O zip original fica em `dados/ans/zip`. Cada mês vira um parquet, para o conjunto inteiro não precisar caber na memória.

O detalhado do país inteiro ocupa vários GB e pode levar horas. Deixe a célula rodando. Se parar no meio, execute de novo: o que já foi gravado não baixa outra vez.

```python
from pathlib import Path

from ans import baixar_tudo

baixar_tudo()

raiz = Path("dados/ans/parquet")
for tipo in ("cons", "det", "rem"):
    arquivos = list((raiz / tipo).glob("*.parquet"))
    mb = sum(arquivo.stat().st_size for arquivo in arquivos) / 1024 / 1024
    print(f"{tipo}: {len(arquivos):,} arquivos, {mb:,.0f} MB")
```
